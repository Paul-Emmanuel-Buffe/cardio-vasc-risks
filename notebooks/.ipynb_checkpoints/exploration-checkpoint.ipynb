### Detection des valeurs atypiques (Outliers)

```python
# Variables critiques pour la précision du modèle: Nous avons répéré plus haut de potenciels outliers
cols_to_audit = ['height', 'weight', 'ap_hi', 'ap_lo']

n = len(cols_to_audit)

plt.figure(figsize=(6 * n, 5))

for i, col in enumerate(cols_to_audit, 1):
    # 1. Visualisation par Boxplot
    plt.subplot(1, n, i)
    sns.boxplot(y=df_patients[col], color="salmon")
    plt.title(f"Visualisation des écarts: {col}")

    # 2. Calcul mathématique de la règle du "bras de levier"
    Q1 = df_patients[col].quantile(0.25)
    Q3 = df_patients[col].quantile(0.75)
    IQR = Q3 - Q1

    limite_basse = Q1 - 1.5 * IQR
    limite_haute = Q3 + 1.5 * IQR

    # Identification des points hors limites (inférieurs à la limite basse OU supérieurs à la limite haute)
    outliers = df_patients[(df_patients[col] < limite_basse) | (df_patients[col] > limite_haute)]
    
    print(f"--- Analyse : {col} ---")
    print(f"Écart Interquartile (IQR) : {IQR:.2f}")
    print(f"Seuil de tolérance bas : {limite_basse:.2f}")  # Ligne ajoutée
    print(f"Seuil de tolérance haut : {limite_haute:.2f}")
    print(f"Nombre de points identifiés : {len(outliers)}\n")

plt.tight_layout()
plt.show()
```

## Nettoyage des données et création de la table patients_cleaned

```python
#### Création de la requette de nettoyage et de création de la nouvelle table

requette_ctas = """
    DROP TABLE IF EXISTS patients_cleaned;
    CREATE TABLE patients_cleaned AS

    WITH calculated_data AS (
    SELECT 
        *,
        ROUND(age / 365.25) AS age_annees,
        ROUND((weight / POWER(height / 100.0, 2))::NUMERIC, 3) AS bmi
    FROM patients
    )

    SELECT
        *,
        ROUND((bmi*cholesterol)::NUMERIC, 3) AS bmi_chol_int
    FROM calculated_data
    WHERE 
        ( 
            (gender = 2 AND height BETWEEN 140 AND 220 AND weight BETWEEN 40 AND 210)
             OR
             (gender = 1 AND height BETWEEN 130 AND 210 AND weight BETWEEN 35 AND 200)
        )

        AND bmi BETWEEN 13 AND 90
        AND ap_hi BETWEEN 40 AND 370
        AND ap_lo BETWEEN 20 AND 190
        AND ap_hi > (ap_lo + 10);
    """

try:
        # Ouverture connection et curseur
        conn = psycopg2.connect(**DB_CONFIG)
        cursor = conn.cursor()

        # Execution de la création de la table avec la requette 'requette_ctas'
        cursor.execute(requette_ctas)

        #Enregistrement dans la BDD
        conn.commit()

        # Comptage du nombre total de lignes dans la table brute 'patients'.
        cursor.execute("SELECT COUNT(*) FROM patients")
        
        # EXPLICATION TECHNIQUE :
        # fetchone() récupère le résultat de la requête SQL sous forme de Tuple (ex: (70000,)).
        # L'utilisation de l'index [0] est indispensable pour "déballer" ce Tuple 
        # et extraire l'entier pur (70000). Sans le [0], les futures opérations 
        # mathématiques (comme la soustraction) planteraient (TypeError).
        nb_total = cursor.fetchone()[0]

        cursor.execute("SELECT COUNT(*) FROM patients_cleaned")
        nb_total_cleaned = cursor.fetchone()[0]

        # Calcul de la diff
        difference = nb_total - nb_total_cleaned

        # Affichage 
        print("-" * 45)
        print("RAPPORT DE NETTOYAGE DES DONNÉES (PIPELINE)")
        print("-" * 45)

        print(f"Nb patients initiaux : {nb_total}.")
        print(f"Patients supprimés (Outliers) : {difference}.")
        print(f"Patients conservés : {nb_total_cleaned}")
        print("-" * 45)

    # -------------------------------------------------------------------------
    # BLOC EXCEPT : GESTION DES ERREURS (Le filet de sécurité)
    # -------------------------------------------------------------------------
except Exception as e:
        # EXPLICATION TECHNIQUE :
        # locals() est un dictionnaire natif de Python qui contient toutes les variables locales actives.
        # On vérifie d'abord si la variable 'conn' a eu le temps d'être créée avant le crash,
        # et on s'assure qu'elle contient bien une connexion active (et non un objet vide/None).
        if 'conn' in locals() and conn:
            
            # ROLLBACK (Principe d'Atomicité) :
            # Si la requête SQL a planté au milieu de son exécution, on annule absolument TOUT.
            # Cela empêche d'avoir une table créée à moitié ou des données corrompues.
            # La base de données revient à l'état exact où elle était avant le 'try'.
            conn.rollback()
            
        # On affiche l'erreur exacte renvoyée par le système pour faciliter le débogage.
        print(f"Erreur lors du nettoyage SQL : {e}")

    # -------------------------------------------------------------------------
    # BLOC FINALLY : NETTOYAGE OBLIGATOIRE DES RESSOURCES
    # -------------------------------------------------------------------------
finally:
        # EXPLICATION TECHNIQUE :
        # Le bloc 'finally' s'exécute TOUJOURS, qu'il y ait eu un succès ou un crash.
        # C'est l'endroit idéal pour "fermer les portes" et éviter les fuites de mémoire 
        # (les fameuses connexions fantômes qui font saturer Docker).
        
        # 1. On ferme d'abord le "messager" (le curseur), s'il a été créé.
        if 'cursor' in locals() and cursor:
            cursor.close()
            
        # 2. On ferme ensuite la connexion principale à la base de données.
        if 'conn' in locals() and conn:
            conn.close()
            print("Ressources PostgreSQL libérées avec succès.")
    
```

### Heapmap

```python
 # 1. Sélection des colonnes pertinentes pour l'analyse de corrélation
columns_to_correlate = [
    'age_annees', 'gender', 'height', 'weight', 'ap_hi', 'ap_lo', 
    'bmi', 'bmi_chol_int', 'smoke', 'alco', 'gluc', 'active', 'cardio', 'cholesterol'
]

# 2. Calcul de la matrice de corrélation (Pearson)
correlation_matrix = df_patients[columns_to_correlate].corr()

# 3. Configuration et affichage de la heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation_matrix, 
    annot=True,           # Affiche les valeurs numériques dans les cases
    cmap='coolwarm',      # Palette de couleurs (bleu = négatif, rouge = positif)
    fmt=".2f",            # Arrondi à 2 décimales
    linewidths=0.5,       # Ajoute une fine ligne de séparation entre les cases
    vmin=-1, vmax=1       # Fixe les bornes de l'échelle des couleurs
)

plt.title("Matrice de corrélation des variables (EDA)", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

#### 1. Analyse des Incidences sur la Variable Cible (cardio)

L'évaluation des coefficients de corrélation avec la variable cible met en évidence trois dynamiques distinctes :

* **Impact majeur :** La pression artérielle systolique (ap_hi : 0,43) et diastolique (ap_lo : 0,34) constituent les prédicteurs linéaires les plus puissants du jeu de données, confirmant l'hypertension comme facteur de risque prédominant.
* **Impact modéré :** L'âge (age_annees : 0,24) indique une progression constante du risque avec le temps. Le cholestérol (cholesterol : 0,22) et l'indice de masse corporelle (bmi : 0,19) présentent également un lien linéaire significatif.
* **Apport du Feature Engineering :** La variable d'interaction bmi_chol_int obtient un coefficient de 0,25. Ce score, supérieur à ceux du bmi seul (0,19) et du cholestérol seul (0,22), valide l'hypothèse selon laquelle l'effet conjugué du surpoids et de la dyslipidémie offre une meilleure puissance prédictive.
* **Impact linéaire négligeable :** La taille (height : -0,01), le glucose (gluc : 0,09) ainsi que les variables comportementales (smoke : -0,02, alco : 0,04, active : 0,01) ne présentent pas de corrélation linéaire directe notable avec la cible.

---

#### 2. Choix de Sélection des Variables et Justifications Mathématiques

Afin de prémunir le modèle de régression logistique contre la multicolinéarité, qui déstabilise le calcul des coefficients, les arbitrages suivants ont été appliqués :

##### Variables exclues du modèle

* **Poids (weight) et Taille (height) :** Le poids presents une colinéarité critique de 0,87 avec le bmi. La taille n'affiche aucun signal direct (-0,01). Le bmi encapsulant mathématiquement ces deux informations tout en offrant une meilleure corrélation avec la cible, le poids et la taille sont retirés par souci de parcimonie.
* **Cholestérol (cholesterol) :** Une colinéarité extrême de 0,93 est mesurée avec la variable d'interaction bmi_chol_int. De plus, le cholestérol affiche une forte corrélation secondaire avec le glucose (0,49). Son exclusion permet d'éliminer ce réseau de redondances massives. La variable combinée bmi_chol_int est donc privilégiée (0,25).

##### Variables conservées dans le modèle

* **Le duo de pressions artérielles (ap_hi et ap_lo) :** Bien que corrélées à 0,73, ce score demeure inférieur au seuil critique standard de 0,80. De plus, elles représentent deux réalités cliniques distinctes et complémentaires en médecine (phases systolique et diastolique). Elles sont donc toutes deux maintenues.
* **Glucose (gluc) :** Malgré sa faible corrélation initiale (0,09), la variable est conservée. L'élimination du cholestérol brut l'affranchit de sa principale colinéarité (0,49), lui permettant d'apporter une information métabolique indépendante dont l'impact non linéaire sera correctement capté après encodage.
* **Les variables comportementales (smoke, alco, active) :** Leurs corrélations proches de zéro indiquent uniquement l'absence de relation linéaire pure. La régression logistique reste pertinente pour capter leurs effets combinés ou des dynamiques non linéaires (effets de seuil). Cliniquement indispensables, leur conservation garantit l'explicabilité médicale du modèle final.

##### Facteurs Comportementaux (Mode de vie)

* **Tabagisme et Alcool (`smoke`, `alco`) :** Les variations de prévalence entre consommateurs et non-consommateurs sont minimes à l'échelle globale. Ces variables sont néanmoins conservées pour permettre au modèle de capter d'éventuels effets cumulatifs non linéaires.
* **Activité Physique (`active`) :** Effet protecteur validé. La proportion de patients atteints de troubles cardiovasculaires diminue de manière visible chez les individus actifs (`active = 1`).

---

##### Variable Démographique (Genre)

* **Genre (`gender`) :** La prévalence de la maladie est rigoureusement identique entre les hommes et les femmes (environ 50 %). L'échantillon présente une asymétrie avec 45 000 femmes pour 25 000 hommes.
* **Décision de modélisation :** La variable est maintenue. L'échantillon minoritaire (25 000 hommes) reste statistiquement massif pour l'apprentissage. Conserver le genre assure un ajustement épidémiologique standard, indispensable pour la validité du modèle, tandis que la régression logistique neutralisera d'elle-même ce coefficient si nécessaire.
